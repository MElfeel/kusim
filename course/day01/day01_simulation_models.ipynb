## 1. Story: the forecast that was "right" for three years

*(An invented but typical situation.)* A young engineer is asked to forecast the production of a new well. She fits a smooth decline curve to the first year of data. For three years the forecast matches reality very well. In year four the well starts producing much more water, and the forecast is wrong by a factor of two.

Was the model bad? Not for the first three years: it was *fit for purpose* during that time. It just did not contain the physics that matters later (water arriving at the well). Today's lesson is a small version of this: a very simple model, solved in two ways, and a careful look at when it can be trusted. Every reservoir simulator is built the same way.

## 2. A simple model: a growing colony

A *simulation model* shows the main features of a real system and is simple enough to calculate with. We use bacteria because the equation is short.

Let $N$ be the number of bacteria. We say the growth rate is proportional to the number present:

$$\frac{dN}{dt} = \alpha N \tag{1}$$

where $\alpha$ is a constant (per hour here). If we start with $N_0$ bacteria at $t=0$, the exact (**analytical**) solution is

$$N(t) = N_0\, e^{\alpha t} \tag{2}$$

> **Math refresher.** $\dfrac{dN}{dt}$ is the *slope* of the curve of $N$ against $t$: how fast bacteria are being added at that moment. Equation (1) says "the more there are, the faster they grow". The number $e \approx 2.718$ is special: the function $e^{x}$ has a slope equal to its own value, which is why it solves equation (1).

Does the model make sense for ever? As $t \to \infty$, $N \to \infty$. A real colony runs out of food, so the model is only good for early growth. We will come back to that.

We use these numbers throughout: $\alpha = 1.74$ per hour, and $N_0 = 100$ bacteria at the start.

## 4. The numerical model: take small steps

Pretend we *cannot* solve equation (1) exactly. We can still step forward in time. Cut time into steps of length $\Delta t$ and write $N^n$ for the value at step $n$ and $N^{n+1}$ for the value one step later. The slope is approximated by the change over one step:

$$\frac{N^{n+1}-N^{n}}{\Delta t} \approx \alpha N^{n} \tag{3}$$

Rearrange to get the unknown $N^{n+1}$ on its own:

$$N^{n+1} = \left(1+\alpha\,\Delta t\right) N^{n} \tag{4}$$

This is an **algorithm** (a recipe): start from $N^0$, apply (4) to get $N^1$, apply it again to get $N^2$, and so on.

### By hand first (5 minutes)

Take $\Delta t = 0.05$ hours, so $1+\alpha\Delta t = 1 + 1.74 \times 0.05 = 1.087$.

| step $n$ | time (h) | $N^n$ (numerical) | $N(t)$ exact |
|---|---|---|---|
| 0 | 0.00 | 100.00 | 100.00 |
| 1 | 0.05 | $100 \times 1.087 = 108.70$ | 109.09 |
| 2 | 0.10 | $108.70 \times 1.087 = 118.16$ | 119.01 |
| 3 | 0.15 | **?** | 129.82 |

Fill in the last row with a calculator. The numerical value is always a little too small: the exact curve bends upward, but each step uses the slope from the *start* of the step.

> **Why this matters for reservoirs.** The saturation update in `kusim` has exactly this shape: *new value = old value + (rate of change) × (step size)*. We come back to it in Section 10.

## 6. Repeating: the `for` loop

Typing `append` 40 times would be tedious (and 2000 times for $\Delta t = 0.001$ impossible). A **`for` loop** repeats a block of code a given number of times.

```python
for step in range(3):
    print("hello")      # this line is indented: it belongs to the loop
```

- `range(3)` gives the numbers 0, 1, 2 (again counting from 0, stopping *before* 3), so the loop runs 3 times.
- The line ends with a colon `:` and the lines that belong to the loop are **indented** (4 spaces). Indentation is how Python knows where the loop ends.
- `step` is a variable that takes each value in turn.

Run this to see it working:

````python
for step in range(4):
    print("this is repetition number", step)
print("loop finished")      # not indented: runs once, after the loop
````

Now the whole numerical solution. We choose $\Delta t$ and the end time, work out how many steps that is, and loop. We also keep a list of times: step $n$ happens at time $n \times \Delta t$.

Read it line by line, then run it.

````python
alpha = 1.74
N0 = 100
dt = 0.05          # time step, hours
t_end = 2.0        # end time, hours

n_steps = round(t_end / dt)         # number of steps (round() gives a whole number)

times = [0.0]
numerical = [N0]

for n in range(1, n_steps + 1):     # n = 1, 2, ..., n_steps
    N_new = (1 + alpha * dt) * numerical[-1]     # equation (4), using the last value
    numerical.append(N_new)
    times.append(n * dt)

print("number of steps :", n_steps)
print("items in the list:", len(numerical))
print(f"final time {times[-1]:.2f} h,  final N = {numerical[-1]:.1f}")
````

Notice that `numerical` has $n_\text{steps}+1$ items: the starting value plus one for each step. The final value is `numerical[-1]`.

The exact answer at $t = 2$ h is $N_0 e^{\alpha t}$. Let us compare:

````python
exact_end = N0 * np.exp(alpha * t_end)
error_percent = abs(numerical[-1] - exact_end) / exact_end * 100

print(f"exact     : {exact_end:.1f}")
print(f"numerical : {numerical[-1]:.1f}")
print(f"error     : {error_percent:.1f} %")
````

## 7. Plotting the two solutions

`plt.plot(x_list, y_list)` draws a curve. For the exact curve we need many points, so we ask NumPy for 200 evenly spaced times with `np.linspace(start, stop, how_many)`. A NumPy **array** is like a list, but a formula such as `np.exp(...)` applies to all of its numbers at once. (We study arrays properly on Day 6.)

````python
t_fine = np.linspace(0, t_end, 200)           # 200 times between 0 and t_end
exact_curve = N0 * np.exp(alpha * t_fine)     # equation (2) at all of them

plt.plot(t_fine, exact_curve, label="analytical (exact)")
plt.plot(times, numerical, "o-", markersize=3, label=f"numerical, dt = {dt}")
plt.xlabel("time (h)")
plt.ylabel("number of bacteria")
plt.legend()
plt.grid(True)
plt.show()
````

The numerical curve sits below the exact one, as predicted by the hand calculation.

## 10. Bridge: the same idea inside `kusim`

Today's recipe was

$$N^{n+1} = N^{n} + \Delta t \cdot \left(\alpha N^{n}\right)$$

*new value = old value + step size × rate of change*.

The reservoir simulator uses the same idea for water saturation. Equation (5-68) of the KUSIM thesis updates the saturation $S_w$ of one grid block explicitly:

$$S_{w}^{\,n+1}=S_{w}^{\,n}+\frac{1}{C_{sw}}\left[\;\sum_{\text{6 faces}} T\left[\left(P_{\text{neighbour}}-P\right)-0.433\,\gamma_o\left(D_{\text{neighbour}}-D\right)\right]\;-\;q_o\;-\;C_{po}\left(P^{\,n+1}-P^{\,n}\right)\right]$$

You are **not expected to understand it today**. Look only at the shape: *new = old + (a sum of terms) divided by a storage coefficient*. The terms are:

| Term | Meaning | Name in the code |
|---|---|---|
| $S_w^{\,n+1}$, $S_w^{\,n}$ | water saturation, new and old | `s_new`, `s_old` |
| $\sum T\,[\dots]$ | oil flowing in from the six neighbouring blocks (pressure difference minus gravity head, times transmissibility) | `flow` and `gra_o` |
| $q_o$ | oil taken out by a well in this block | `qop_cell` |
| $C_{po}\,(P^{n+1}-P^{n})$ | oil released or stored because the pressure changed | `cpoo * (p_new - p_old)` |
| $C_{sw}$ | storage coefficient: how much a change in saturation matters (it also contains $\Delta t$ and the pore volume) | `cswo` |

Notice how the **step size** is hidden inside $C_{sw}$. Like our bacteria, a larger $\Delta t$ means bigger jumps, and past some limit the solution oscillates. Day 8 is about exactly that.

*(Optional peek.)* The same equation written in Python in `src/kusim/simulator.py` is a single line. The names in the table above appear in it, and the code keeps the gravity term `gra_o` separate from the pressure part:

```python
s_new = s_old + (flow - qop_cell - cpoo * (p_new - p_old) - gra_o) / cswo
```

(`/` is divide, `*` is multiply, as you learnt in Section 3.2.)
